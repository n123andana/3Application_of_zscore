## 2)Outlier Detection:

```python
import numpy as np
from scipy.stats import zscore

z_scores = zscore(df['Maths'])

threshold = 3

outliers = df[np.abs(z_scores) > threshold]

print("Outliers:")
print(outliers[['Unnamed: 0', 'Maths']])
```

 no outlier for the threshold of 3

```python
threshold=2
outliers = df[np.abs(z_scores) > threshold]

print("Outliers:")
print(outliers[['Unnamed: 0', 'Maths']])
```

```python
#using IQR
Q1 = df['Maths'].quantile(0.25)
Q3 = df['Maths'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df['Maths'] < lower) |
    (df['Maths'] > upper)
]

print("Lower limit:", lower)
print("Upper limit:", upper)
print("Outliers:")
print(outliers[['Unnamed: 0', 'Maths']])
```

```python
import numpy as np
from scipy.stats import zscore

z_scores = zscore(df['Civics'])

threshold = 3

outliers = df[np.abs(z_scores) > threshold]

print("Outliers:")
print(outliers["Civics"])
```

```python
subjects = [
    'Maths', 'Physics', 'Chemistry', 'English',
    'Biology', 'Economics', 'History', 'Civics'
]

for col in subjects:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\n{col} outliers:")
    print(outliers[['Unnamed: 0', col]])
```

```python
from sklearn.preprocessing import StandardScaler

data = df[subjects]
scaler = StandardScaler()
scaler.fit(data)
```

```python
print("Mean:", scaler.mean_)
print("Variance:", scaler.var_)
```

```python
scaled_data = scaler.transform(data)
scaled_data
```

```python
scaled_data = scaler.transform(data)

scaled_df = pd.DataFrame(scaled_data, columns=subjects)

scaled_df
```

### 1)probability using zscore
probability that Sepal Length is less than 5 cm

```python
from scipy.stats import norm

mean = df['SepalLengthCm'].mean()
std = df['SepalLengthCm'].std()

x = 5

z = (x - mean) / std
print("Z-score:", z)

probability = norm.cdf(z)
print("Probability:", probability)
```

```python
#to find the probability of Sepal Length being greater than 5 cm:
probability = 1 - norm.cdf(z)
print("Probability:", probability)
```
